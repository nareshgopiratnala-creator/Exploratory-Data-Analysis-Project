```python

print(df.groupby('Property_Type')['Price'].agg(['mean', 'median', 'count']))
print(df.groupby('Location')['Price'].agg(['mean', 'median', 'count']))
print(df.groupby('Has_Pool')['Price'].mean())
print(df.groupby('Furnishing')['Price'].mean())
```

```python
fig, axes = plt.subplots(2, 2, figsize=(18, 11))

#Property Price Distribution chart 1
sns.histplot(
    df["Price"],
     ax=axes[0, 0], color="royalblue", bins=30
)
axes[0, 0].set_title(
    "Property Price Distribution", fontsize=13, fontweight="bold"
)
axes[0, 0].set_xlabel("Price")
axes[0, 0].set_ylabel("Count")
# Price Vs Area Chart 2
sns.scatterplot(
    data=df,
    x="Area_SqFt",
    y=df["Price"],
    hue="Property_Type",
    alpha=0.7,
    ax=axes[0, 1],
)
axes[0, 1].set_title(
    "Price vs. Area (SqFt)", fontsize=13, fontweight="bold"
)
axes[0, 1].set_xlabel("Area (SqFt)")
axes[0, 1].set_ylabel("Price")

sns.barplot(
    data=df,
    x="Property_Type",
    y=df["Price"] ,
    hue="Property_Type",
    legend=False,
    ax=axes[1, 0],
)
axes[1, 0].set_title(
    "Price by Property Type", fontsize=13, fontweight="bold"
)
axes[1, 0].set_xlabel("Property Type")
axes[1, 0].set_ylabel("Price ")

cityorder = (
    df.groupby("Location")["Price"].mean().sort_values(ascending=False).index
)
sns.barplot(
    data=df,
    x="Location",
    y=df["Price"],
    order=cityorder,
    hue="Location",
    legend=False,
    errorbar=None,
    ax=axes[1, 1],
)
axes[1, 1].set_title(
    "Average Price by City", fontsize=13, fontweight="bold"
)
axes[1, 1].set_xlabel("Location")
axes[1, 1].set_ylabel("Average Price")


plt.tight_layout()
plt.show()
```
